```python
import numpy as np
import pandas as pd
import os

print("Librerías cargadas correctamente.")
```

```python
np.random.seed(42)
n = 1_000_000

df = pd.DataFrame({
    "edad": np.random.normal(40, 12, n),
    "ingresos": np.random.lognormal(10, 0.5, n),
    "altura": np.random.normal(170, 10, n),
    "experiencia": np.random.randint(0, 41, n),
    "categoria": np.random.choice(["A", "B", "C", "D", "E"], n)
})

print("Dataset de 1M generado.")
```

```python
print("Dimensiones:", df.shape) # Debería mostrar (1000000, 5)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df["categoria"].value_counts()
```

```python
df.to_parquet(
    "../data/dataset_1M.parquet",
    index=False
)
print("dataset_1M.parquet guardado .")
```
